```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
dataset.head()
```

```python
dataset.info()
```

## Data Cleaning

```python
 ## Missing values
dataset.isnull().sum()
```

```python
#Missing values in the dataset .rows having missing values can be displayed using the following code:
dataset[dataset.isnull().any(axis=1)]
```

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122:,"Region"]=1
df=dataset
```

```python
df.info()
```

```python
df[['Region']]=df[['Region']].astype('int')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## Removing the null values #Remove bad/missing rows → Fix numbering → Save result in df.
df=df.dropna().reset_index(drop=True)
df.head()
```

```python
df.iloc[[122]]
```

```python
## Remove the 122nd row from the dataset and reset the index
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[122]
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## fix spaces in columns names,used to remove extra spaces from all column names in a DataFrame.
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

## Change the required columns as integer datatype

```python
df[['month','day','year','Temperature','RH','Ws']]=df[['month','day','year','Temperature','RH','Ws']].astype('int')
```

```python
df.info()
```

## Changing the other columns to float datatype

```python
objects=[features for features in df.columns if df[features].dtypes=='O']
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
    
```

```python
df.info()
```

```python
df.describe()
```

```python
##Let save the cleaned dataset
df.to_csv('Algerian_forest_fires_cleaned_dataset.csv',index=False)
```

## Exploratory Data Analysis

```python
##axis=0 → Remove rows (vertical direction).
##axis=1 → Remove columns (horizontal direction)
## drop day,month,year
df_copy=df.drop(['day','month','year'],axis=1)
```

```python
df_copy.head()
```

```python
## Encoding of the categories in classes,if classes is not fire then 0 else 1
df_copy['Classes']=np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy.head()
```

```python
# Count of values in the Classes column
df_copy['Classes'].value_counts()
```

```python
# Count of values in the Classes column
df_copy['Classes'].value_counts(normalize=True)*100
```

# Visualization

```python
##sns.set_theme() → Applies Seaborn's default theme for better-looking plots.
##plt.style.use('seaborn-v0_8') → Sets the Seaborn plotting style in Matplotlib.
##df_copy.hist(bins=50, figsize=(20,15)) → Creates histograms for all numerical columns with 50 intervals (bins).
##plt.show() → Displays all the plots.
## Plot density plot for all features
sns.set_theme()
plt.style.use('seaborn-v0_8')
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
#df_copy['Classes'] → Selects the Classes column.
#value_counts() → Counts how many times each class appears.
#normalize=True → Converts counts into proportions (0 to 1).
#*100 → Converts proportions into percentages
## Percentage for Pie Chart
percentage=df_copy['Classes'].value_counts(normalize=True)*100
```

```python
# Plotting Piechart
classlabels=["Fire","Not Fire"]
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')
plt.title("Pie Chart of Classes")
plt.show()
```
